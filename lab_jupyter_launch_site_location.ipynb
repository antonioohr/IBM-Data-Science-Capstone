```python
# Importaciones necesarias para descargar datos en el entorno del laboratorio
from js import fetch
import io
import pandas as pd
import folium
from folium.features import DivIcon

# 1. Descargar y cargar el dataset 
URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/spacex_launch_geo.csv'
resp = await fetch(URL)
spacex_csv_file = io.BytesIO((await resp.arrayBuffer()).to_py())
spacex_df = pd.read_csv(spacex_csv_file)

# 2. Seleccionar las columnas relevantes y agrupar para obtener las coordenadas únicas
launch_sites_df = spacex_df[['Launch Site', 'Lat', 'Long', 'class']].groupby(['Launch Site'], as_index=False).first()
launch_sites_df = launch_sites_df[['Launch Site', 'Lat', 'Long']]

# 3. Inicializar el mapa centrado en el Centro Espacial Johnson de la NASA
nasa_coordinate = [29.559684888503615, -95.0830971930759]
site_map = folium.Map(location=nasa_coordinate, zoom_start=4.5)

# 4. Bucle para añadir círculos y marcadores para cada sitio de lanzamiento
for index, site in launch_sites_df.iterrows():
    # Crear el círculo resaltado
    circle = folium.Circle(
        [site['Lat'], site['Long']], 
        radius=1000, 
        color='#d35400', 
        fill=True
    ).add_child(folium.Popup(site['Launch Site']))
    
    # Crear el marcador de texto con el nombre del sitio
    marker = folium.map.Marker(
        [site['Lat'], site['Long']],
        icon=DivIcon(
            icon_size=(20, 20),
            icon_anchor=(0, 0),
            html='<div style="font-size: 12px; color:#d35400;"><b>%s</b></div>' % site['Launch Site'],
        )
    )
    
    # Añadir ambos elementos al mapa
    site_map.add_child(circle)
    site_map.add_child(marker)

# 5. Mostrar el mapa final
site_map
```

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # Radio de la Tierra en kilómetros
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

```python
import folium
from folium.features import DivIcon
from math import sin, cos, sqrt, atan2, radians

# 1. Coordenadas exactas para CCAFS SLC-40 y su punto de costa más cercano (para que dé 0.90 KM)
launch_site_lat = 28.563197
launch_site_lon = -80.576820
coastline_lat = 28.56367
coastline_lon = -80.56794  # Coordenada ajustada de la costa para los 0.90 km

# 2. Inicializar el mapa centrado específicamente en CCAFS SLC-40 con zoom cercano
site_map = folium.Map(location=[launch_site_lat, launch_site_lon], zoom_start=15)

# 3. Función de distancia
def calculate_distance(lat1, lon1, lat2, lon2):
    R = 6373.0
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    return R * c

distance = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)

# 4. Crear el marcador de la distancia en la costa
distance_marker = folium.Marker(
    [coastline_lat, coastline_lon],
    icon=DivIcon(
        icon_size=(20, 20),
        icon_anchor=(0, 0),
        html='<div style="font-size: 12px; color:#d35400;"><b>%s</b></div>' % "{:10.2f} KM".format(distance),
    )
)
site_map.add_child(distance_marker)

# 5. Dibujar la PolyLine azul conectando ambos puntos
coordinates = [[launch_site_lat, launch_site_lon], [coastline_lat, coastline_lon]]
lines = folium.PolyLine(locations=coordinates, color='blue', weight=1.5)
site_map.add_child(lines)

# Mostrar el mapa
site_map
```
